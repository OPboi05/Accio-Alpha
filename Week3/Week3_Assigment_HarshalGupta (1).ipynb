```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)  # for reproducibility
n_steps = 50        # number of time steps
vel = 10  # constant velocity
dt = 1.0           # time step
```

```python
true_positions = []
measurements = []
position = 0.0
measurement_noise_std = 2 # sensor noise (R)
for _ in range(n_steps):
    position += vel*dt
    true_positions.append(position)
    
    
    noisy_measurement = position + np.random.normal(0, measurement_noise_std)
    measurements.append(noisy_measurement)
```

```python
# Initial state estimate (position only)
x_est = np.array([[1]])  # initial guess

# Initial uncertainty
P = np.array([10])     # large value = low confidence

# State transition matrix
F = np.array([[1.0]])

# Measurement matrix
H = np.array([[1.0]])

# Process noise (model uncertainty)
Q = np.array([[0.1]])

# Measurement noise (sensor uncertainty)
R = np.array([[measurement_noise_std ** 2]])
```

```python
estimates = []
for z in measurements:
    # ------------------
    # PREDICTION STEP
    # ------------------
    x_pred = F @ x_est[0]
    P_pred = F @ P @ F.T + Q

    # ------------------
    # UPDATE STEP
    # ------------------
    S = H @ P_pred @ H.T + R
    K = P_pred @ H.T @ np.linalg.inv(S)

    y = np.array([[z]]) - H @ x_pred   # innovation
    x_est = x_pred + K @ y

    P = (np.eye(1) - K @ H) @ P_pred

    estimates.append(x_est[0, 0])
```

```python
plt.figure(figsize=(10, 5))
plt.plot(true_positions, label="True Position", linewidth=2)

plt.scatter(range(n_steps), measurements, label="Noisy Measurements", color="red", alpha=0.6)
plt.plot(estimates, label="Kalman Filter Estimate", linewidth=2)
plt.xlabel("Time Step")
plt.ylabel("Position")
plt.title("1D Kalman Filter: Position Estimation")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)  # for reproducibility
n_steps = 50        # number of time steps
vel = 100  # constant velocity
sigma_a = 25 #St. Deviation
dt = 1.0           # time step
```

```python
true_positions = []
true_velocities = []
measurements = []
position = 0.0
measurement_noise_std = 2.0 # sensor noise (R)
for _ in range(n_steps):
    a = np.random.normal(0, sigma_a)
    position += vel * dt + 0.5*a*dt**2
    true_positions.append(position)
    vel += a*dt
    true_velocities.append(vel)
    noisy_measurement = position + np.random.normal(0, measurement_noise_std)
    measurements.append(noisy_measurement)
```

```python
# Initial state estimate (position only)
x_est = np.array([[1,1000]])  # initial guess

# Initial uncertainty
P = np.array([[2, 0], [0,2]])     # large value = low confidence

# State transition matrix
F = np.array([[1.0, dt], [0.0, 1.0]])

# Measurement matrix
H = np.array([[1.0, 0.0]])

# Process noise (model uncertainty)
Q = np.array([[0.1 , 0.2], [0.2, 0.1]])

# Measurement noise (sensor uncertainty)
R = np.array([[measurement_noise_std ** 2]])
```

```python
estimates_x = []
estimates_y = []
for z in measurements:
    # ------------------
    # PREDICTION STEP
    # ------------------
    x_pred = F @ x_est[0]
    P_pred = F @ P @ F.T + Q

    # ------------------
    # UPDATE STEP
    # ------------------
    S = H @ P_pred @ H.T + R
    K = P_pred @ H.T @ np.linalg.inv(S)

    y = np.array([[z]]) - H @ x_pred   # innovation
    x_est = x_pred + K @ y

    P = (np.eye(1) - K @ H) @ P_pred

    estimates_x.append(x_est[0,0])
    estimates_y.append(x_est[0,1])
```

```python
plt.figure(figsize=(10, 5))
plt.plot(true_positions, label="True Position", linewidth=2)
plt.plot(true_velocities, label="True Velocities", linewidth=2)
plt.scatter(range(n_steps), measurements, label="Noisy Measurements", color="red", alpha=0.6)
plt.plot(estimates_x, label="Kalman Filter Estimate(For X)", linewidth=2)
plt.plot(estimates_y, label="Kalman Filter Estimate(For V)", linewidth=2)
plt.xlabel("Time Step")
plt.ylabel("Position")
plt.title("1D Kalman Filter: Position Estimation")
plt.legend()
plt.grid(True)

plt.show()
```
